## Set Font Properties for Title and Labels

fontdict parameter in xlabel(), ylabel(), and title() to set font properties for the title and labels.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("hhhh", fontdict = font1)
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.show()
```

## Matplotlib Subplots
What if we need more than one graph, in a single plot.
Which means, subplots.
We can use subplots() function.


```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([12,33,44,55])
y=np.array([12,34,55,67])
plt.subplot(1,2,1)
plt.plot(x,y)
x=np.array([14,56,77,89])
y=np.array([14,44,55,66])
plt.subplot(1,2,2)#here the plot of second parameters is asked to be given in second column graph
plt.plot(x,y)
plt.show()
```

```python
x=np.array([12,33,44,55])
y=np.array([12,34,55,67])
plt.subplot(1,2,1)
plt.plot(x,y)
x=np.array([14,56,77,89])
y=np.array([14,44,55,66])
plt.subplot(1,2,1)#here in(1,2,1):1-row,2-column,1-where it should be ploted
plt.plot(x,y)
plt.show()
```

plt.subplot(1, 2, 1)
plt.subplot(1, 2, 2)

This two lines here represents the figure and the plots in the figure, which means this 2 lines will decide  how many subplots are there and where it need to allocate.
1,2,1 = for 1st subplot
1,2,2 = for 2nd subplot
If we type 1,2,3 there will an error, because It calculated like this:1 * 2 = 2
Maximum subplot will be 2

## Size
You can change the size of the dots with the s argument.
Just like colors, make sure the array for sizes has the same length as the arrays for the x- and y-axis:


```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])
plt.scatter(x, y, s=sizes)
plt.show()
```

##Alpha
you can adjust the transparency of the dots with the alpha argument.
Just like colors, make sure the array for sizes has the same length as the arrays for the x- and y-axis




```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])
plt.scatter(x, y, s=sizes, alpha=0.3)
plt.show()

```

## Combine Color Size and Alpha


```python
x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = 10 * np.random.randint(100, size=(100))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')
plt.colorbar()
plt.show()
```

## Legend
To add a list of explanation for each wedge, use the legend() functio

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
plt.pie(y, labels = mylabels)
plt.legend()
plt.show()
```

##Legend With Header
To add a header to the legend, add the title parameter to the legend function

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels = mylabels)
plt.legend(title = "Four Fruits:")
plt.show()
```

```python
from pywaffle import Waffle
import matplotlib.pyplot as plt
```

```python
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    columns=10,
    values=[48, 46, 6],
    figsize=(5, 3)
)
plt.show()
```

```python
data = {'Democratic': 48, 'Republican': 46, 'Libertarian': 3}
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    values=data,
    legend={'loc': 'upper left', 'bbox_to_anchor': (1.1, 1)}
)
plt.show()
```

```python
data = {'Democratic': 48, 'Republican': 46, 'Libertarian': 3}
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    values=data,
    colors=["#232066", "#983D3D", "#DCB732"],
    title={'label': 'Vote Percentage in 2016 US Presidential Election', 'loc': 'left'},
    labels=[f"{k} ({v}%)" for k, v in data.items()],
    legend={'loc': 'lower left', 'bbox_to_anchor': (0, -0.4), 'ncol': len(data), 'framealpha': 0},
    starting_location='NW',
    block_arranging_style='snake'
)
fig.set_facecolor('#EEEEEE')
plt.show()
```

```python

# Plot with Icons - Pictogram Chart
data = {'Democratic': 48, 'Republican': 46, 'Libertarian': 3}
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    values=data,
    colors=["#232066", "#983D3D", "#DCB732"],
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)},
    icons='child',
    font_size=12,
    icon_legend=True
)
plt.show()
```
