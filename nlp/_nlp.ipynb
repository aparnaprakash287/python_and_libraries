🔹 NOUNS

NN (Noun, singular or mass)
Names a person, place, thing, or idea (singular).
book, water, city

NNS (Noun, plural)
Plural form of a noun.
books, cities

NNP (Proper noun, singular)
Specific name of a person/place/organization.
India, John, Google

NNPS (Proper noun, plural)
Plural proper nouns.
Americans, Alps

🔹 VERBS

VB (Verb, base form)
Basic form of a verb.
eat, run

VBD (Verb, past tense)
Action that happened in the past.
ate, ran

VBG (Verb, gerund/present participle)
Verb ending in -ing, used as verb or noun.
running, eating

VBN (Verb, past participle)
Used with auxiliary verbs.
eaten, broken

VBP (Verb, present tense, non-3rd person singular)
Present tense verb except he/she/it.
I run, they eat

VBZ (Verb, present tense, 3rd person singular)
Present tense for he/she/it.
runs, eats

🔹 ADJECTIVES

JJ (Adjective)
Describes a noun.
big, happy

JJR (Adjective, comparative)
Compares two things.
bigger, happier

JJS (Adjective, superlative)
Compares more than two things.
biggest, happiest

🔹 ADVERBS

RB (Adverb)
Modifies a verb, adjective, or another adverb.
quickly, very

RBR (Adverb, comparative)
Comparison form of adverb.
faster

RBS (Adverb, superlative)
Highest degree of adverb.
fastest

🔹 PRONOUNS

PRP (Personal pronoun)
Replaces a noun.
I, you, he, she, they

PRP$ (Possessive pronoun)
Shows ownership.
my, your, his

WP (Wh-pronoun)
Used in questions or clauses.
who, what

WP$ (Possessive wh-pronoun)
Shows ownership in questions.
whose

🔹 DETERMINERS

DT (Determiner)
Introduces a noun.
a, an, the

PDT (Predeterminer)
Comes before a determiner.
all, both

WDT (Wh-determiner)
Used in questions.
which

🔹 CONJUNCTIONS & PREPOSITIONS

CC (Coordinating conjunction)
Joins words or clauses.
and, but, or

IN (Preposition / subordinating conjunction)
Shows relation or introduces clauses.
in, on, because

🔹 NUMBERS

CD (Cardinal number)
Numbers or quantities.
one, 5, ten

🔹 OTHER IMPORTANT TAGS

TO – The word to
Used before verbs.
to go

MD (Modal verb)
Expresses possibility, ability, necessity.
can, should, must

UH (Interjection)
Expresses emotion.
oh, wow

EX (Existential “there”)
Indicates existence.
There is a book.

FW (Foreign word)
Word from another language.
bon appétit

LS (List item marker)
Used in lists.
1., a., i.

POS (Possessive ending)
Shows ownership.
John’s

RP (Particle)
Part of a phrasal verb.
give up, take off

SYM (Symbol)
Mathematical or special symbols.
+, %, &

$ – Dollar sign

. – Sentence-ending punctuation

, – Comma

: – Colon or ellipsis

**Chunking**

```python
import nltk
text = "The quick brown fox jumps over the lazy dog"
tokens = nltk.word_tokenize(text)
tags = nltk.pos_tag(tokens)
chunk_grammar = "NP: {<DT>?<JJ>*<NN>}"
cp = nltk.RegexpParser(chunk_grammar)
tree = cp.parse(tags)
tree.pretty_print()
```

 **TF-IDF**

- TF = Term Frequency → How often a word appears in a document.

- IDF = Inverse Document Frequency → Measures how important a word is across all documents.

 - Words that appear in many documents get down-weighted.

- TF-IDF = TF × IDF → Higher value means the word is important for that document but rare across others.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer()
X=tf.fit_transform(["This is a sample","This is another example"])
print(tf.get_feature_names_out())
print(X.toarray())
```

| Term             | Simple meaning                     |
| ---------------- | ---------------------------------- |

| Term             | Range                      | Meaning                                      |
| ---------------- | -------------------------- | -------------------------------------------- |
| **Polarity**     | `-1.0` to `+1.0`           | How **negative** or **positive** the text is |
|                  | `-1.0` → Very negative     | Example: *“I hate this movie.”*              |
|                  | `0.0` → Neutral            | Example: *“It is a movie.”*                  |
|                  | `+1.0` → Very positive     | Example: *“I love this movie!”*              |
|                  |                            |                                              |

|                  | `0.0` → Completely factual | Example: *“The sun rises in the east.”*      |
|                  | `1.0` → Fully opinionated  | Example: *“I think this is the best ever!”*  |


# Text Classification with Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
texts=["Buy now",
       "Limited offer",
       "Call me later",
       "Meeting at noon",
       "Lets have lunch",
       "Win a price",
       "Project iscussion tomorrow",
       "Free coupons"]
labels=["spam",
        "spam",
        "ham",
        "ham",
        "spam",
        "ham",
        "spam",
        "ham"]
X_train,X_test,y_train,y_test=train_test_split(
    texts,labels,test_size=0.5,random_state=42#test_size should be 0 to 1
)
pipe=Pipeline([("Vectorizer",CountVectorizer()),
               ("Classifier",MultinomialNB())
               ])
pipe.fit(X_train,y_train)
preds=pipe.predict(X_test)
print("Test predictions:",preds)
print("Actual labels: ",y_test)
print("Accuracy:",accuracy_score(y_test,preds))
new_text=["Congratulations,you won a free ticket!"]
new_pred=pipe.predict(new_text)
print("\nNew text:",new_text[0])
print("Predicted label:",new_pred[0])
```
