## Concat
It is used to concat two or more dataframes

```python
import pandas as pd
```

```python
tcs={'id':[101,102,103,104],'name':['Mithun','Dipin','Jose','Rahul'],
     'Age':[25,26,27,29]}

wipro={'id':[101,102,103,104],
       'name':['dev','sreenath','sreerag','shafeel'],
       'age':[24,24,27,26]}
tcs_emp=pd.DataFrame(tcs)
tcs_emp

```

```python
wipro_emp=pd.DataFrame(wipro)
wipro_emp
```

```python
pd.concat([tcs_emp,wipro_emp])
```

```python
#ignore index-To get common index after concatenation
pd.concat([tcs_emp,wipro_emp],ignore_index=True)
```

```python
#keys-To provide seperate keys for each concatenate data frames
data=pd.concat([tcs_emp,wipro_emp],keys=["tcs","wipro"])
data
```

```python
#concatenating vertically
data=pd.concat([tcs_emp,wipro_emp],axis=1)
data
```

```python
#cocatinating a dataframe with a series
salary=pd.Series([65000,78000,30000],name="salary")
pd.concat([tcs_emp,salary],axis=1)
```

## Merge
Merge is a process to merge two tables horizontally

```python
tcs={'id':[101,102,103,104],'name':['Mithun','Dipin','Jose','Rahul'],
     'age':[25,26,27,29]}

wipro={'id':[101,102,103,104],
       'name':['dev','sreenath','sreerag','shafeel'],
       'age':[24,24,27,26]}

tcs_emp=pd.DataFrame(tcs)
wipro_emp=pd.DataFrame(wipro)

tcs_emp
```

```python
wipro_emp
```

```python
pd.merge(tcs_emp,wipro_emp,on="id")
```

```python
pd.merge(tcs_emp,wipro_emp,on="id",suffixes=["left","rigght"])
```

```python
pd.merge(tcs_emp,wipro_emp,on="id",how="outer")#outer join
```

```python
pd.merge(tcs_emp,wipro_emp,on="age",how="outer")#merge on the basis of age and displays the result having same age
```

##Cross tab or contingency table

In pandas, pd.crosstab() is used to compute a cross-tabulation of two or more factors. It’s often used for analyzing categorical variables, similar to a contingency table in statistics.

syntax:

pd.crosstab(index, columns, values=None, aggfunc=None, margins=False, margins_name='All')


```python
df=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/dataset/survey_crosstab.csv")
df
```

```python
#Create cros  tabulation
ct=pd.crosstab(df["Gender"],df["Preferred_Device"])
print(ct)
```

```python
# Cross-tab of AgeGroup vs Preferred_Device with totals:
ct=pd.crosstab(df["AgeGroup"],df["Preferred_Device"],margins=True,margins_name="Total")
print(ct)
```
